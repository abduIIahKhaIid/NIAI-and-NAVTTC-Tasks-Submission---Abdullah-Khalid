## Text Processing with NLTK

## Objective
Perform common text-processing operations with NLTK: tokenization, normalization, stop-word filtering, stemming, frequency analysis, and bigram generation.

This notebook avoids corpus downloads by using NLTK components that do not require external datasets.

```python
from nltk.tokenize import wordpunct_tokenize
from nltk.stem import PorterStemmer
from nltk import FreqDist
from nltk.util import bigrams

text = '''
Artificial Intelligence is transforming software development.
Machine learning models learn patterns from data, and good text processing
helps natural language systems work with raw text.
'''
```

## Tokenization and Normalization

```python
tokens = wordpunct_tokenize(text.lower())
words = [token for token in tokens if token.isalpha()]

print(words)
```

## Stop-Word Filtering

```python
stop_words = {"is", "from", "and", "with", "the", "a", "an", "to", "of", "in"}
filtered_words = [word for word in words if word not in stop_words]
print(filtered_words)
```

## Stemming

```python
stemmer = PorterStemmer()
stems = [stemmer.stem(word) for word in filtered_words]

for original, stem in zip(filtered_words, stems):
    print(f"{original:15s} -> {stem}")
```

## Frequency Distribution

```python
freq = FreqDist(filtered_words)
print(freq.most_common(10))
```

## Bigrams

```python
pairs = list(bigrams(filtered_words))
print(pairs[:10])
```

## Conclusion
The notebook demonstrates a compact NLTK text-processing pipeline without requiring downloadable corpora.
