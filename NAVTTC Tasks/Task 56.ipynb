## Demonstration of Neural Networks from Scratch

## Objective
Build a small feed-forward neural network using only **NumPy** so the forward pass, binary cross-entropy loss, backpropagation, and gradient-descent updates are visible.

```python
import numpy as np

# XOR dataset
X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=float)
y = np.array([[0],[1],[1],[0]], dtype=float)

rng = np.random.default_rng(42)
W1 = rng.normal(0, 0.5, size=(2, 4))
b1 = np.zeros((1, 4))
W2 = rng.normal(0, 0.5, size=(4, 1))
b2 = np.zeros((1, 1))
```

## Activation Function

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(a):
    return a * (1 - a)
```

## Training with Forward and Backward Passes

```python
learning_rate = 0.5
epochs = 10000
eps = 1e-8

for epoch in range(epochs):
    # Forward pass
    z1 = X @ W1 + b1
    a1 = sigmoid(z1)
    z2 = a1 @ W2 + b2
    y_hat = sigmoid(z2)

    # Binary cross-entropy
    loss = -np.mean(y*np.log(y_hat + eps) + (1-y)*np.log(1-y_hat + eps))

    # Backpropagation
    dz2 = (y_hat - y) / len(X)
    dW2 = a1.T @ dz2
    db2 = np.sum(dz2, axis=0, keepdims=True)

    da1 = dz2 @ W2.T
    dz1 = da1 * sigmoid_derivative(a1)
    dW1 = X.T @ dz1
    db1 = np.sum(dz1, axis=0, keepdims=True)

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    if epoch % 2000 == 0:
        print(f"Epoch {epoch:5d} | Loss = {loss:.6f}")
```

## Final Predictions

```python
hidden = sigmoid(X @ W1 + b1)
output = sigmoid(hidden @ W2 + b2)
predictions = (output >= 0.5).astype(int)

print("Probabilities:")
print(np.round(output, 4))
print("Predicted:", predictions.ravel())
print("Actual   :", y.astype(int).ravel())
```

## Conclusion
The network learns the XOR relationship using one hidden layer. The task exposes the major computational steps behind a basic neural network.
