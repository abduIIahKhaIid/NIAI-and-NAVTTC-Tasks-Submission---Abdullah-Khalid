## Neural Network Loss Functions

## Objective
Understand and calculate several loss functions used in neural networks. A loss function converts prediction error into a scalar quantity that an optimizer attempts to minimize.

```python
import numpy as np
eps = 1e-8
```

## Mean Squared Error (MSE)
MSE is common in regression and squares each prediction error before averaging.

```python
y_true = np.array([3.0, 5.0, 2.5, 7.0])
y_pred = np.array([2.5, 5.5, 2.0, 6.0])

mse = np.mean((y_true - y_pred) ** 2)
print("MSE:", mse)
```

## Mean Absolute Error (MAE)

```python
mae = np.mean(np.abs(y_true - y_pred))
print("MAE:", mae)
```

## Binary Cross-Entropy

```python
y_true_binary = np.array([1, 0, 1, 1], dtype=float)
y_prob_binary = np.array([0.9, 0.2, 0.7, 0.6], dtype=float)

bce = -np.mean(
    y_true_binary * np.log(y_prob_binary + eps)
    + (1 - y_true_binary) * np.log(1 - y_prob_binary + eps)
)
print("Binary cross-entropy:", round(bce, 6))
```

## Categorical Cross-Entropy

```python
y_true_cat = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1]
], dtype=float)

y_prob_cat = np.array([
    [0.8, 0.1, 0.1],
    [0.2, 0.7, 0.1],
    [0.1, 0.2, 0.7]
], dtype=float)

cce = -np.mean(np.sum(y_true_cat * np.log(y_prob_cat + eps), axis=1))
print("Categorical cross-entropy:", round(cce, 6))
```

## Hinge Loss

```python
y_true_hinge = np.array([1, -1, 1, -1], dtype=float)
scores = np.array([1.2, -0.5, 0.3, 0.4], dtype=float)

hinge = np.mean(np.maximum(0, 1 - y_true_hinge * scores))
print("Hinge loss:", round(hinge, 6))
```

## Choosing a Loss
- **MSE/MAE** are common for regression.
- **Binary cross-entropy** is appropriate for binary probabilistic classification.
- **Categorical cross-entropy** is common for multiclass softmax outputs.
- **Hinge loss** is associated with margin-based classification.

The correct loss depends on the target representation and model output.
