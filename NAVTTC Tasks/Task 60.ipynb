## Linguistics / NLP Using Machine Learning

## Objective
Build a basic text-classification pipeline using **TF-IDF** features and **Logistic Regression**. The example shows how raw language is converted into numerical features for machine learning.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

texts = [
    "I love this course and the lessons are excellent",
    "The class is useful and very interesting",
    "This project is amazing and easy to follow",
    "The teacher explains concepts clearly",
    "I dislike this lesson and it is confusing",
    "The material is boring and difficult",
    "This course is terrible and badly organized",
    "I am unhappy with the explanation",
    "The examples are practical and helpful",
    "The assignment is clear and well structured",
    "The lecture is confusing and too difficult",
    "I hate the way this topic is explained",
]

labels = [
    "positive", "positive", "positive", "positive",
    "negative", "negative", "negative", "negative",
    "positive", "positive", "negative", "negative"
]

X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.33, random_state=42, stratify=labels
)
```

## Build the Pipeline

```python
model = Pipeline([
    ("tfidf", TfidfVectorizer(lowercase=True, ngram_range=(1, 2))),
    ("classifier", LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)
predictions = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, predictions), 4))
print(classification_report(y_test, predictions))
```

## Predict New Text

```python
new_sentences = [
    "The lesson was extremely helpful",
    "This topic is badly explained and confusing"
]

for sentence, prediction in zip(new_sentences, model.predict(new_sentences)):
    print(f"{sentence!r} -> {prediction}")
```

## Conclusion
TF-IDF converts vocabulary patterns into numerical features, while logistic regression learns the classification boundary from those features.
