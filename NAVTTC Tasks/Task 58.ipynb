## Feed-Forward Neural Network

## Objective
Demonstrate a feed-forward neural network using TensorFlow/Keras. Information moves from the input layer through hidden layers to the output layer without recurrent feedback connections.

```python
import numpy as np
import tensorflow as tf
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

tf.random.set_seed(42)
np.random.seed(42)

iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Define the Architecture

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(3, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

## Train and Evaluate

```python
history = model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=16,
    verbose=0
)

test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=0)
print("Test loss:", round(test_loss, 4))
print("Test accuracy:", round(test_accuracy, 4))
```

## Make Predictions

```python
probabilities = model.predict(X_test[:5], verbose=0)
predicted_classes = probabilities.argmax(axis=1)

for i, (pred, actual) in enumerate(zip(predicted_classes, y_test[:5]), start=1):
    print(f"Sample {i}: predicted={iris.target_names[pred]}, actual={iris.target_names[actual]}")
```

## Conclusion
This notebook demonstrates the architecture, training, evaluation, and prediction workflow of a standard feed-forward neural network.
